```python
import os
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import joblib
import yaml
import h5py

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)

sns.set_theme(style="whitegrid")


# ==============================================================
# 2. DEFINE PROJECT PATHS
# ==============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Mining Performance Rating Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_265_AU_1557_A_to_D.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)


def output_path(filename):

    return os.path.join(
        BASE_DIR,
        "pso_" + filename
    )


# ==============================================================
# 3. SAVE AND DISPLAY GRAPH FUNCTION
# ==============================================================

def save_show_graph(filename):

    path = output_path(filename)

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print(f"\n[SAVED] {path}")

    plt.show()

    plt.close()


# ==============================================================
# 4. LOAD DATASET
# ==============================================================

print("\n" + "=" * 65)
print("LOADING MINING PERFORMANCE DATASET")
print("=" * 65)

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())


# ==============================================================
# 5. DATA PREPROCESSING
# ==============================================================

print("\nPreprocessing Dataset...")

if df.shape[1] < 3:

    raise ValueError(
        "Dataset must contain at least three columns."
    )

# Select first three columns
df = df.iloc[:, :3].copy()

df.columns = [
    "Year",
    "Three_Star_Mines",
    "Five_Star_Mines"
]

# Convert numeric columns
for col in [
    "Three_Star_Mines",
    "Five_Star_Mines"
]:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Extract financial year
df["Year_Numeric"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year_Numeric"] = pd.to_numeric(
    df["Year_Numeric"],
    errors="coerce"
)

# Remove missing records
df = df.dropna(
    subset=[
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
).copy()

df["Year_Numeric"] = (
    df["Year_Numeric"].astype(int)
)

# Sort chronologically
df = df.sort_values(
    "Year_Numeric"
).reset_index(drop=True)

# Validate dataset
if len(df) < 3:

    raise ValueError(
        "At least three observations are required."
    )

if df["Year_Numeric"].duplicated().any():

    raise ValueError(
        "Duplicate financial years detected."
    )

if (
    (df["Three_Star_Mines"] < 0).any()
    or (df["Five_Star_Mines"] < 0).any()
):

    raise ValueError(
        "Mine counts cannot be negative."
    )

if (
    df["Five_Star_Mines"]
    > df["Three_Star_Mines"]
).any():

    raise ValueError(
        "Five-star mines cannot exceed "
        "three-star-and-above mines."
    )

# Calculate five-star percentage
df["Five_Star_Percentage"] = (
    df["Five_Star_Mines"]
    / df["Three_Star_Mines"].replace(0, np.nan)
) * 100

df["Five_Star_Percentage"] = (
    df["Five_Star_Percentage"].fillna(0)
)

# Calculate year-on-year growth
df["Three_Star_Growth"] = (
    df["Three_Star_Mines"]
    .pct_change() * 100
)

df["Five_Star_Growth"] = (
    df["Five_Star_Mines"]
    .pct_change() * 100
)

print("\nProcessed Dataset:")
print(df)

df.to_csv(
    output_path("processed_data.csv"),
    index=False
)


# ==============================================================
# 6. PREPARE FEATURES AND TARGETS
# ==============================================================

X = df[
    ["Year_Numeric"]
].values.astype(float)

y = df[
    [
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
].values.astype(float)

years = df["Year"].astype(str).values

target_names = [
    "Three-Star and Above Mines",
    "Five-Star Mines"
]


# ==============================================================
# 7. NORMALIZE INPUT AND TARGET VALUES
# ==============================================================

print("\nNormalizing Dataset...")

BASE_YEAR = int(X.min())

# Normalize years
x_normalized = (
    X.flatten() - BASE_YEAR
).astype(float)

# Normalize targets
y_mean = np.mean(
    y,
    axis=0
)

y_std = np.std(
    y,
    axis=0
)

y_std = np.where(
    y_std == 0,
    1.0,
    y_std
)

y_scaled = (
    y - y_mean
) / y_std


# ==============================================================
# 8. PSO CONFIGURATION
# ==============================================================

PSO_CONFIG = {

    "population_size": 80,

    "iterations": 500,

    "inertia_weight": 0.7,

    "cognitive_coefficient": 1.5,

    "social_coefficient": 1.5,

    "polynomial_degree": 1,

    "position_min": -5.0,

    "position_max": 5.0,

    "velocity_max": 0.5,

    "random_seed": 42
}

POPULATION_SIZE = (
    PSO_CONFIG["population_size"]
)

ITERATIONS = (
    PSO_CONFIG["iterations"]
)

W = (
    PSO_CONFIG["inertia_weight"]
)

C1 = (
    PSO_CONFIG["cognitive_coefficient"]
)

C2 = (
    PSO_CONFIG["social_coefficient"]
)

DEGREE = (
    PSO_CONFIG["polynomial_degree"]
)

POSITION_MIN = (
    PSO_CONFIG["position_min"]
)

POSITION_MAX = (
    PSO_CONFIG["position_max"]
)

VELOCITY_MAX = (
    PSO_CONFIG["velocity_max"]
)


# ==============================================================
# 9. GENERATE POLYNOMIAL FEATURES
# ==============================================================

def polynomial_features(x, degree=1):

    x = np.asarray(
        x,
        dtype=float
    ).reshape(-1)

    features = [
        np.ones_like(x)
    ]

    for power in range(
        1,
        degree + 1
    ):

        features.append(
            x ** power
        )

    return np.column_stack(
        features
    )


X_poly = polynomial_features(
    x_normalized,
    DEGREE
)

NUM_FEATURES = (
    X_poly.shape[1]
)

NUM_TARGETS = (
    y_scaled.shape[1]
)

NUM_PARAMETERS = (
    NUM_FEATURES * NUM_TARGETS
)


# ==============================================================
# 10. PARTICLE REPRESENTATION
# ==============================================================

# Each particle represents the coefficients of
# a multi-output regression model.
#
# With degree = 1:
#
# Three-Star Mines = a1 + b1*x
# Five-Star Mines  = a2 + b2*x
#
# PSO optimizes the regression coefficients.


def decode_particle(particle):

    return np.asarray(
        particle,
        dtype=float
    ).reshape(
        NUM_FEATURES,
        NUM_TARGETS
    )


# ==============================================================
# 11. PARTICLE PREDICTION FUNCTION
# ==============================================================

def particle_predict(
    particle,
    features
):

    coefficients = decode_particle(
        particle
    )

    predictions = (
        features @ coefficients
    )

    return predictions


# ==============================================================
# 12. PSO FITNESS FUNCTION
# ==============================================================

# Lower MSE indicates better fitness.
#
# L2 regularization discourages
# unnecessarily large coefficients.

def fitness_function(particle):

    predictions = particle_predict(
        particle,
        X_poly
    )

    mse = np.mean(
        (y_scaled - predictions) ** 2
    )

    regularization = (
        0.0001
        * np.mean(
            np.asarray(particle) ** 2
        )
    )

    return float(
        mse + regularization
    )


# ==============================================================
# 13. PARTICLE SWARM OPTIMIZATION CLASS
# ==============================================================

class ParticleSwarmOptimization:

    def __init__(self):

        self.rng = np.random.default_rng(
            PSO_CONFIG["random_seed"]
        )

        self.population_size = (
            POPULATION_SIZE
        )

        self.iterations = (
            ITERATIONS
        )

        self.dimensions = (
            NUM_PARAMETERS
        )

        self.best_position = None

        self.best_fitness = np.inf

        self.history = []


    # ----------------------------------------------------------
    # INITIALIZE PARTICLES
    # ----------------------------------------------------------

    def initialize_particles(self):

        positions = self.rng.uniform(
            POSITION_MIN,
            POSITION_MAX,
            size=(
                self.population_size,
                self.dimensions
            )
        )

        velocities = self.rng.uniform(
            -VELOCITY_MAX,
            VELOCITY_MAX,
            size=(
                self.population_size,
                self.dimensions
            )
        )

        return positions, velocities


    # ----------------------------------------------------------
    # EVALUATE PARTICLES
    # ----------------------------------------------------------

    def evaluate_particles(self, positions):

        fitness_values = np.array(
            [
                fitness_function(particle)
                for particle in positions
            ]
        )

        return fitness_values


    # ----------------------------------------------------------
    # TRAIN PSO
    # ----------------------------------------------------------

    def fit(self):

        print("\n" + "=" * 65)
        print("TRAINING PARTICLE SWARM OPTIMIZATION")
        print("=" * 65)

        positions, velocities = (
            self.initialize_particles()
        )

        # Evaluate initial population
        fitness_values = (
            self.evaluate_particles(
                positions
            )
        )

        # Personal best positions
        personal_best_positions = (
            positions.copy()
        )

        personal_best_fitness = (
            fitness_values.copy()
        )

        # Global best particle
        best_index = np.argmin(
            fitness_values
        )

        global_best_position = (
            positions[best_index].copy()
        )

        global_best_fitness = float(
            fitness_values[best_index]
        )

        # ------------------------------------------------------
        # MAIN PSO LOOP
        # ------------------------------------------------------

        for iteration in range(
            self.iterations
        ):

            # Generate random coefficients
            r1 = self.rng.random(
                size=positions.shape
            )

            r2 = self.rng.random(
                size=positions.shape
            )

            # PSO velocity update
            velocities = (
                W * velocities

                + C1 * r1 * (
                    personal_best_positions
                    - positions
                )

                + C2 * r2 * (
                    global_best_position
                    - positions
                )
            )

            # Limit velocity
            velocities = np.clip(
                velocities,
                -VELOCITY_MAX,
                VELOCITY_MAX
            )

            # Update particle positions
            positions = (
                positions + velocities
            )

            # Apply position boundaries
            positions = np.clip(
                positions,
                POSITION_MIN,
                POSITION_MAX
            )

            # Evaluate updated particles
            fitness_values = (
                self.evaluate_particles(
                    positions
                )
            )

            # Update personal best
            improved = (
                fitness_values
                < personal_best_fitness
            )

            personal_best_positions[
                improved
            ] = positions[
                improved
            ].copy()

            personal_best_fitness[
                improved
            ] = fitness_values[
                improved
            ]

            # Find best personal position
            best_index = np.argmin(
                personal_best_fitness
            )

            current_best_fitness = float(
                personal_best_fitness[
                    best_index
                ]
            )

            # Update global best
            if (
                current_best_fitness
                < global_best_fitness
            ):

                global_best_fitness = (
                    current_best_fitness
                )

                global_best_position = (
                    personal_best_positions[
                        best_index
                    ].copy()
                )

            # Record optimization history
            self.history.append(
                {
                    "Iteration":
                        iteration + 1,

                    "Best_Fitness":
                        global_best_fitness,

                    "Mean_Fitness":
                        float(
                            np.mean(
                                fitness_values
                            )
                        )
                }
            )

            # Display progress
            if (
                iteration + 1
            ) % 50 == 0:

                print(
                    f"Iteration {iteration + 1}"
                    f" | Best Fitness: "
                    f"{global_best_fitness:.8f}"
                )

        self.best_position = (
            global_best_position
        )

        self.best_fitness = (
            global_best_fitness
        )

        print("\nPSO Training Completed.")

        print(
            "Best Fitness:",
            self.best_fitness
        )

        return self


# ==============================================================
# 14. TRAIN PSO MODEL
# ==============================================================

pso = ParticleSwarmOptimization()

pso.fit()


# ==============================================================
# 15. CREATE PSO REGRESSION MODEL
# ==============================================================

# This model uses PSO-optimized regression coefficients.


def predict_with_coefficients(
    years_input,
    coefficients,
    base_year,
    degree,
    target_mean,
    target_std
):

    years_input = np.asarray(
        years_input,
        dtype=float
    ).reshape(-1)

    normalized_years = (
        years_input - base_year
    )

    features = polynomial_features(
        normalized_years,
        degree
    )

    scaled_predictions = (
        features @ coefficients
    )

    predictions = (
        scaled_predictions
        * target_std
        + target_mean
    )

    return predictions


best_coefficients = decode_particle(
    pso.best_position
)


def pso_predict(years_input):

    return predict_with_coefficients(
        years_input,
        best_coefficients,
        BASE_YEAR,
        DEGREE,
        y_mean,
        y_std
    )


# ==============================================================
# 16. SAVE PSO MODEL AS PKL
# ==============================================================

print("\nSaving PSO Model as PKL...")

# Save a dictionary rather than a custom Python class.
# This makes the model easier to load in other scripts.

model_data = {

    "algorithm":
        "Particle Swarm Optimization",

    "model_type":
        "Multi-Output Polynomial Regression",

    "coefficients":
        best_coefficients,

    "base_year":
        BASE_YEAR,

    "polynomial_degree":
        DEGREE,

    "target_mean":
        y_mean,

    "target_std":
        y_std,

    "best_fitness":
        pso.best_fitness,

    "feature_names": [
        "Year_Numeric"
    ],

    "target_names": [
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
}

joblib.dump(
    model_data,
    output_path("model.pkl")
)

print(
    "[SAVED]",
    output_path("model.pkl")
)


# ==============================================================
# 17. SAVE PSO MODEL AS H5
# ==============================================================

# H5 stores PSO model parameters.
# It is not a TensorFlow/Keras neural network.

print("\nSaving PSO Model as H5...")

with h5py.File(
    output_path("model.h5"),
    "w"
) as h5:

    h5.create_dataset(
        "coefficients",
        data=best_coefficients
    )

    h5.create_dataset(
        "target_mean",
        data=y_mean
    )

    h5.create_dataset(
        "target_std",
        data=y_std
    )

    h5.create_dataset(
        "best_particle",
        data=pso.best_position
    )

    h5.attrs["base_year"] = (
        BASE_YEAR
    )

    h5.attrs["polynomial_degree"] = (
        DEGREE
    )

    h5.attrs["best_fitness"] = (
        pso.best_fitness
    )

    h5.attrs["algorithm"] = (
        "Particle Swarm Optimization"
    )

print(
    "[SAVED]",
    output_path("model.h5")
)


# ==============================================================
# 18. GENERATE HISTORICAL PREDICTIONS
# ==============================================================

print("\nGenerating Historical Predictions...")

pso_predictions = pso_predict(
    X.flatten()
)

print("\nPSO Historical Predictions:")
print(pso_predictions)


# ==============================================================
# 19. CALCULATE PSO TRAINING METRICS
# ==============================================================

print("\nCalculating PSO Training Metrics...")

metrics_records = []

for i, target in enumerate(target_names):

    actual = y[:, i]

    predicted = pso_predictions[:, i]

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mse
    )

    r2 = r2_score(
        actual,
        predicted
    )

    metrics_records.append(
        {
            "Algorithm": "PSO",

            "Target": target,

            "MAE": float(mae),

            "MSE": float(mse),

            "RMSE": float(rmse),

            "R2_Score": float(r2)
        }
    )

metrics_df = pd.DataFrame(
    metrics_records
)

print("\nPSO Training Metrics:")
print(metrics_df)

metrics_df.to_csv(
    output_path("metrics.csv"),
    index=False
)


# ==============================================================
# 20. PSO ACCURACY GRAPH
# ==============================================================

print("\nGenerating PSO Accuracy Graph...")

# R² is a regression fit metric, not classification accuracy.

plt.figure(
    figsize=(11, 6)
)

bars = plt.bar(
    metrics_df["Target"],
    metrics_df["R2_Score"],
    color=["royalblue", "darkorange"]
)

plt.title(
    "PSO Mining Model Training Fit (R²)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category"
)

plt.ylabel(
    "Training R² Score"
)

plt.axhline(
    0,
    color="black",
    linestyle="--"
)

for bar, value in zip(
    bars,
    metrics_df["R2_Score"]
):

    plt.annotate(
        f"{value:.4f}",
        xy=(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center"
    )

plt.grid(
    axis="y",
    alpha=0.3
)

save_show_graph(
    "accuracy_graph.png"
)


# ==============================================================
# 21. PSO CORRELATION HEATMAP
# ==============================================================

print("\nGenerating PSO Heatmap...")

heatmap_data = df[
    [
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines",
        "Five_Star_Percentage"
    ]
]

correlation_matrix = (
    heatmap_data.corr()
)

plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "PSO Mining Performance Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.yticks(
    rotation=0
)

save_show_graph(
    "heatmap.png"
)


# ==============================================================
# 22. PSO COMPARISON GRAPH
# ==============================================================

print("\nGenerating PSO Comparison Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

for i, target in enumerate(target_names):

    axes[i].plot(
        years,
        y[:, i],
        marker="o",
        linewidth=3,
        label="Actual"
    )

    axes[i].plot(
        years,
        pso_predictions[:, i],
        marker="s",
        linestyle="--",
        linewidth=2,
        label="PSO Predicted"
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "Actual vs PSO Predicted Mining Performance",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "comparison_graph.png"
)


# ==============================================================
# 23. GENERATE PSO RESULT CSV
# ==============================================================

print("\nGenerating PSO Result CSV...")

result_df = pd.DataFrame(
    {
        "Year": df["Year"],

        "Actual_Three_Star_Mines":
            y[:, 0].astype(int),

        "PSO_Three_Star_Prediction":
            pso_predictions[:, 0],

        "Actual_Five_Star_Mines":
            y[:, 1].astype(int),

        "PSO_Five_Star_Prediction":
            pso_predictions[:, 1]
    }
)

result_df["PSO_Three_Star_Error"] = np.abs(
    result_df["Actual_Three_Star_Mines"]
    - result_df["PSO_Three_Star_Prediction"]
)

result_df["PSO_Five_Star_Error"] = np.abs(
    result_df["Actual_Five_Star_Mines"]
    - result_df["PSO_Five_Star_Prediction"]
)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nPSO Result CSV:")
print(result_df)


# ==============================================================
# 24. PSO RESULT GRAPH
# ==============================================================

print("\nGenerating PSO Result Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

x_positions = np.arange(
    len(years)
)

for i, target in enumerate(target_names):

    axes[i].bar(
        x_positions - 0.2,
        y[:, i],
        width=0.4,
        label="Actual"
    )

    axes[i].bar(
        x_positions + 0.2,
        pso_predictions[:, i],
        width=0.4,
        label="PSO Predicted"
    )

    axes[i].set_xticks(
        x_positions
    )

    axes[i].set_xticklabels(
        years
    )

    axes[i].set_title(
        target
    )

    axes[i].set_xlabel(
        "Financial Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

fig.suptitle(
    "PSO Mining Performance Results",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "result_graph.png"
)


# ==============================================================
# 25. GENERATE FUTURE PREDICTIONS
# ==============================================================

print("\nGenerating PSO Future Predictions...")

last_year = int(
    df["Year_Numeric"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
)

future_predictions_raw = pso_predict(
    future_years
)


# ==============================================================
# 26. CLEAN FUTURE PREDICTIONS
# ==============================================================

def clean_predictions(predictions):

    predictions = np.asarray(
        predictions,
        dtype=float
    ).copy()

    # Prevent negative counts
    predictions = np.maximum(
        predictions,
        0
    )

    # Round to whole mines
    predictions = np.rint(
        predictions
    ).astype(int)

    # Five-star mines must not exceed
    # three-star-and-above mines.

    predictions[:, 1] = np.minimum(
        predictions[:, 1],
        predictions[:, 0]
    )

    return predictions


future_predictions = clean_predictions(
    future_predictions_raw
)


# ==============================================================
# 27. CREATE PSO PREDICTION CSV
# ==============================================================

future_labels = [
    f"{year}-{str((year + 1) % 100).zfill(2)}"
    for year in future_years
]

prediction_df = pd.DataFrame(
    {
        "Year": future_labels,

        "PSO_Three_Star_Mines":
            future_predictions[:, 0],

        "PSO_Five_Star_Mines":
            future_predictions[:, 1]
    }
)

prediction_df["PSO_Five_Star_Percentage"] = (
    prediction_df["PSO_Five_Star_Mines"]
    / prediction_df[
        "PSO_Three_Star_Mines"
    ].replace(0, np.nan)
) * 100

prediction_df[
    "PSO_Five_Star_Percentage"
] = prediction_df[
    "PSO_Five_Star_Percentage"
].fillna(0)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nPSO Future Predictions:")
print(prediction_df)


# ==============================================================
# 28. PSO PREDICTION GRAPH
# ==============================================================

print("\nGenerating PSO Prediction Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(17, 7)
)

historical_years = (
    df["Year_Numeric"].values
)

for i, target in enumerate(target_names):

    # Historical actual values
    axes[i].plot(
        historical_years,
        y[:, i],
        marker="o",
        linewidth=3,
        label="Historical Actual"
    )

    # Future PSO predictions
    axes[i].plot(
        future_years,
        future_predictions[:, i],
        marker="s",
        linestyle="--",
        linewidth=2,
        label="PSO Forecast"
    )

    # Connect historical and future values
    axes[i].plot(
        [
            historical_years[-1],
            future_years[0]
        ],
        [
            y[-1, i],
            future_predictions[0, i]
        ],
        linestyle="--",
        alpha=0.5
    )

    # Forecast boundary
    axes[i].axvline(
        historical_years[-1],
        linestyle=":",
        color="gray",
        label="Forecast Boundary"
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year Starting Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "PSO Mining Performance Future Forecast "
    "(Illustrative Only)",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "prediction_graph.png"
)


# ==============================================================
# 29. PSO ERROR GRAPH
# ==============================================================

print("\nGenerating PSO Error Graph...")

plt.figure(
    figsize=(11, 6)
)

sns.barplot(
    data=metrics_df,
    x="Target",
    y="MAE",
    color="teal"
)

plt.title(
    "PSO Training Mean Absolute Error",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category"
)

plt.ylabel(
    "Mean Absolute Error (Mine Count)"
)

plt.grid(
    axis="y",
    alpha=0.3
)

save_show_graph(
    "error_graph.png"
)


# ==============================================================
# 30. PSO CONVERGENCE GRAPH
# ==============================================================

print("\nGenerating PSO Convergence Graph...")

history_df = pd.DataFrame(
    pso.history
)

history_df.to_csv(
    output_path("training_history.csv"),
    index=False
)

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    history_df["Iteration"],
    history_df["Best_Fitness"],
    linewidth=2,
    label="Global Best Fitness"
)

plt.plot(
    history_df["Iteration"],
    history_df["Mean_Fitness"],
    linewidth=1.5,
    alpha=0.6,
    label="Mean Swarm Fitness"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Iteration"
)

plt.ylabel(
    "Fitness (Lower Is Better)"
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_show_graph(
    "convergence_graph.png"
)


# ==============================================================
# 31. SAVE PSO YAML CONFIGURATION
# ==============================================================

print("\nSaving PSO Configuration...")

yaml_config = {

    "project": {
        "name":
            "AI-Based Mining Performance Rating Analysis",

        "algorithm":
            "Particle Swarm Optimization",

        "dataset":
            os.path.basename(DATASET_PATH)
    },

    "pso_parameters":
        PSO_CONFIG,

    "model": {
        "type":
            "Multi-Output Polynomial Regression",

        "optimization":
            "Particle Swarm Optimization",

        "model_file":
            "pso_model.pkl",

        "h5_file":
            "pso_model.h5"
    },

    "dataset": {
        "records":
            int(len(df)),

        "features": [
            "Year_Numeric"
        ],

        "targets": [
            "Three_Star_Mines",
            "Five_Star_Mines"
        ]
    },

    "forecast": {
        "years": 5,

        "file":
            "pso_prediction.csv"
    },

    "evaluation": {
        "type":
            "In-sample training evaluation",

        "validated_test_accuracy":
            False
    }
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False
    )


# ==============================================================
# 32. SAVE PSO JSON METADATA
# ==============================================================

print("\nSaving PSO Metadata...")

metadata = {

    "project_name":
        "AI-Based Mining Performance Rating Analysis",

    "algorithm":
        "Particle Swarm Optimization",

    "dataset_records":
        int(len(df)),

    "year_range": {
        "start":
            int(df["Year_Numeric"].min()),

        "end":
            int(df["Year_Numeric"].max())
    },

    "pso_parameters":
        PSO_CONFIG,

    "best_fitness":
        float(pso.best_fitness),

    "best_particle":
        pso.best_position.tolist(),

    "model_coefficients":
        best_coefficients.tolist(),

    "training_metrics":
        metrics_df.to_dict(
            orient="records"
        ),

    "future_predictions":
        prediction_df.to_dict(
            orient="records"
        ),

    "generated_files": [
        "pso_model.pkl",
        "pso_model.h5",
        "pso_accuracy_graph.png",
        "pso_heatmap.png",
        "pso_comparison_graph.png",
        "pso_result.csv",
        "pso_result_graph.png",
        "pso_prediction.csv",
        "pso_prediction_graph.png",
        "pso_error_graph.png",
        "pso_convergence_graph.png"
    ],

    "limitations":
        (
            "Only three annual observations are available. "
            "Reported metrics are in-sample training metrics. "
            "Future forecasts are illustrative extrapolations."
        )
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )


# ==============================================================
# 33. VERIFY SAVED PSO MODEL
# ==============================================================

print("\nVerifying Saved PSO Model...")

loaded_model = joblib.load(
    output_path("model.pkl")
)

loaded_prediction = predict_with_coefficients(
    [2027],
    loaded_model["coefficients"],
    loaded_model["base_year"],
    loaded_model["polynomial_degree"],
    loaded_model["target_mean"],
    loaded_model["target_std"]
)

print(
    "\nTest Prediction for Financial Year 2027-28:"
)

print(
    clean_predictions(loaded_prediction)
)


# ==============================================================
# 34. VERIFY ALL GENERATED FILES
# ==============================================================

print("\n" + "=" * 65)
print("VERIFYING PSO OUTPUT FILES")
print("=" * 65)

generated_files = [

    "model.pkl",
    "model.h5",

    "config.yaml",
    "metadata.json",

    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",

    "result.csv",
    "result_graph.png",

    "prediction.csv",
    "prediction_graph.png",

    "error_graph.png",
    "convergence_graph.png",

    "training_history.csv",
    "metrics.csv",

    "processed_data.csv"
]

for filename in generated_files:

    full_filename = (
        "pso_" + filename
    )

    path = os.path.join(
        BASE_DIR,
        full_filename
    )

    if os.path.exists(path):

        print(
            "[SUCCESS]",
            full_filename
        )

    else:

        print(
            "[FAILED]",
            full_filename
        )


# ==============================================================
# 35. FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 70)

print(
    "AI-BASED MINING PERFORMANCE RATING ANALYSIS"
)

print(
    "USING PARTICLE SWARM OPTIMIZATION"
)

print("=" * 70)

print("\nAlgorithm:")
print("Particle Swarm Optimization")

print("\nPSO Parameters:")

print(
    "Population Size:",
    POPULATION_SIZE
)

print(
    "Iterations:",
    ITERATIONS
)

print(
    "Inertia Weight:",
    W
)

print(
    "Cognitive Coefficient:",
    C1
)

print(
    "Social Coefficient:",
    C2
)

print("\nBest Fitness:")
print(pso.best_fitness)

print("\nGenerated Graphs:")

print("1. pso_accuracy_graph.png")
print("2. pso_heatmap.png")
print("3. pso_comparison_graph.png")
print("4. pso_result_graph.png")
print("5. pso_prediction_graph.png")
print("6. pso_error_graph.png")
print("7. pso_convergence_graph.png")

print("\nGenerated CSV Files:")

print("1. pso_result.csv")
print("2. pso_prediction.csv")
print("3. pso_metrics.csv")
print("4. pso_training_history.csv")
print("5. pso_processed_data.csv")

print("\nGenerated Model Files:")

print("1. pso_model.pkl")
print("2. pso_model.h5")

print("\nGenerated Configuration Files:")

print("1. pso_config.yaml")
print("2. pso_metadata.json")

print("\nOutput Directory:")
print(BASE_DIR)

print(
    "\nWARNING: Only three annual observations "
    "are available. Training-fit scores "
    "are not validated forecast accuracy."
)

print("\nPSO PROJECT EXECUTION COMPLETED SUCCESSFULLY!")

print("=" * 70)
```
