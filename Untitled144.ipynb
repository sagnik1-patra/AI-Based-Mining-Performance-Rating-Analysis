```python
import os
import json
import warnings
from dataclasses import dataclass, asdict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import joblib
import yaml
import h5py

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)


# ==============================================================
# 2. DEFINE PROJECT PATHS
# ==============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Mining Performance Rating Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_265_AU_1557_A_to_D.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)


def output_path(filename):

    return os.path.join(
        BASE_DIR,
        "ais_" + filename
    )


# ==============================================================
# 3. SAVE AND DISPLAY GRAPH FUNCTION
# ==============================================================

def save_show_graph(filename):

    path = output_path(filename)

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        f"\n[SAVED] {path}"
    )

    plt.show()

    plt.close()


# ==============================================================
# 4. LOAD DATASET
# ==============================================================

print("\n" + "=" * 65)
print("LOADING MINING PERFORMANCE DATASET")
print("=" * 65)

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())


# ==============================================================
# 5. DATA PREPROCESSING
# ==============================================================

print("\nPreprocessing Dataset...")

if df.shape[1] < 3:

    raise ValueError(
        "Dataset must contain at least three columns."
    )

# First three columns represent:
# Financial Year
# Three-star and above mines
# Five-star mines

df = df.iloc[:, :3].copy()

df.columns = [
    "Year",
    "Three_Star_Mines",
    "Five_Star_Mines"
]

# Convert mining counts to numeric values
for col in [
    "Three_Star_Mines",
    "Five_Star_Mines"
]:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Extract starting financial year
df["Year_Numeric"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year_Numeric"] = pd.to_numeric(
    df["Year_Numeric"],
    errors="coerce"
)

# Remove incomplete records
df = df.dropna(
    subset=[
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
).copy()

df["Year_Numeric"] = (
    df["Year_Numeric"].astype(int)
)

# Sort by financial year
df = df.sort_values(
    "Year_Numeric"
).reset_index(drop=True)

# Validate data
if len(df) < 3:

    raise ValueError(
        "At least three observations are required."
    )

if df["Year_Numeric"].duplicated().any():

    raise ValueError(
        "Duplicate financial years detected."
    )

if (
    (df["Three_Star_Mines"] < 0).any()
    or (df["Five_Star_Mines"] < 0).any()
):

    raise ValueError(
        "Mine counts cannot be negative."
    )

if (
    df["Five_Star_Mines"]
    > df["Three_Star_Mines"]
).any():

    raise ValueError(
        "Five-star mines cannot exceed "
        "three-star-and-above mines."
    )

# Calculate five-star percentage
df["Five_Star_Percentage"] = (
    df["Five_Star_Mines"]
    / df["Three_Star_Mines"].replace(0, np.nan)
) * 100

df["Five_Star_Percentage"] = (
    df["Five_Star_Percentage"].fillna(0)
)

# Year-on-year growth
df["Three_Star_Growth"] = (
    df["Three_Star_Mines"]
    .pct_change() * 100
)

df["Five_Star_Growth"] = (
    df["Five_Star_Mines"]
    .pct_change() * 100
)

print("\nProcessed Dataset:")
print(df)

df.to_csv(
    output_path("processed_data.csv"),
    index=False
)


# ==============================================================
# 6. PREPARE FEATURES AND TARGETS
# ==============================================================

X = df[
    ["Year_Numeric"]
].values.astype(float)

y = df[
    [
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
].values.astype(float)

years = df["Year"].astype(str).values

target_names = [
    "Three-Star and Above Mines",
    "Five-Star Mines"
]


# ==============================================================
# 7. FEATURE NORMALIZATION
# ==============================================================

# Use the first year as the reference year.
# This prevents very large numerical values.

BASE_YEAR = int(X.min())

x_normalized = (
    X.flatten() - BASE_YEAR
).astype(float)

# Scale targets separately to balance AIS optimization.

y_mean = np.mean(
    y,
    axis=0
)

y_std = np.std(
    y,
    axis=0
)

# Prevent division by zero
y_std = np.where(
    y_std == 0,
    1.0,
    y_std
)

y_scaled = (
    y - y_mean
) / y_std


# ==============================================================
# 8. AIS MODEL CONFIGURATION
# ==============================================================

@dataclass
class AISConfig:

    population_size: int = 80

    generations: int = 500

    elite_size: int = 15

    clones_per_antibody: int = 10

    mutation_rate: float = 0.25

    mutation_decay: float = 0.995

    random_injection: int = 10

    polynomial_degree: int = 1

    random_seed: int = 42


config = AISConfig()


# ==============================================================
# 9. POLYNOMIAL FEATURE GENERATION
# ==============================================================

def polynomial_features(x, degree=1):

    x = np.asarray(
        x,
        dtype=float
    ).reshape(-1)

    features = [
        np.ones_like(x)
    ]

    for power in range(1, degree + 1):

        features.append(
            x ** power
        )

    return np.column_stack(features)


X_poly = polynomial_features(
    x_normalized,
    degree=config.polynomial_degree
)

NUM_FEATURES = X_poly.shape[1]

NUM_TARGETS = y_scaled.shape[1]

NUM_PARAMETERS = (
    NUM_FEATURES * NUM_TARGETS
)


# ==============================================================
# 10. ANTIBODY REPRESENTATION
# ==============================================================

# Each antibody represents the coefficients of a
# multi-output regression model.
#
# For degree=1:
#
# y1 = a1 + b1*x
# y2 = a2 + b2*x
#
# AIS optimizes:
# [a1, b1, a2, b2]


def decode_antibody(antibody):

    return np.asarray(
        antibody,
        dtype=float
    ).reshape(
        NUM_FEATURES,
        NUM_TARGETS
    )


# ==============================================================
# 11. AIS PREDICTION FUNCTION
# ==============================================================

def antibody_predict(
    antibody,
    features
):

    coefficients = decode_antibody(
        antibody
    )

    predictions = (
        features @ coefficients
    )

    return predictions


# ==============================================================
# 12. FITNESS FUNCTION
# ==============================================================

# Lower MSE indicates better fitness.
#
# A small L2 regularization term discourages
# unnecessarily large model coefficients.

def fitness_function(antibody):

    predictions = antibody_predict(
        antibody,
        X_poly
    )

    mse = np.mean(
        (y_scaled - predictions) ** 2
    )

    regularization = (
        0.0001
        * np.mean(
            np.asarray(antibody) ** 2
        )
    )

    return float(
        mse + regularization
    )


# ==============================================================
# 13. CLONAL SELECTION AIS IMPLEMENTATION
# ==============================================================

class ArtificialImmuneSystem:

    def __init__(self, config):

        self.config = config

        self.rng = np.random.default_rng(
            config.random_seed
        )

        self.best_antibody = None

        self.best_fitness = np.inf

        self.history = []


    # ----------------------------------------------------------
    # INITIALIZE ANTIBODY POPULATION
    # ----------------------------------------------------------

    def initialize_population(self):

        population = self.rng.normal(
            loc=0.0,
            scale=1.0,
            size=(
                self.config.population_size,
                NUM_PARAMETERS
            )
        )

        return population


    # ----------------------------------------------------------
    # EVALUATE POPULATION
    # ----------------------------------------------------------

    def evaluate_population(self, population):

        fitness_values = np.array(
            [
                fitness_function(antibody)
                for antibody in population
            ]
        )

        return fitness_values


    # ----------------------------------------------------------
    # CLONING OPERATION
    # ----------------------------------------------------------

    def clone_antibodies(self, elites):

        clones = []

        for antibody in elites:

            for _ in range(
                self.config.clones_per_antibody
            ):

                clones.append(
                    antibody.copy()
                )

        return np.array(clones)


    # ----------------------------------------------------------
    # HYPERMUTATION
    # ----------------------------------------------------------

    def hypermutate(
        self,
        clones,
        elite_fitness,
        generation
    ):

        mutated_clones = []

        # Mutation gradually decreases
        decay = (
            self.config.mutation_decay
            ** generation
        )

        for i, clone in enumerate(clones):

            elite_index = (
                i // self.config.clones_per_antibody
            )

            parent_fitness = elite_fitness[
                elite_index
            ]

            # Higher fitness error produces
            # stronger mutation.

            relative_error = (
                parent_fitness
                / (1.0 + parent_fitness)
            )

            mutation_strength = (
                self.config.mutation_rate
                * decay
                * (0.2 + relative_error)
            )

            noise = self.rng.normal(
                loc=0.0,
                scale=mutation_strength,
                size=clone.shape
            )

            mutated = (
                clone + noise
            )

            mutated_clones.append(
                mutated
            )

        return np.array(
            mutated_clones
        )


    # ----------------------------------------------------------
    # SELECTION OPERATION
    # ----------------------------------------------------------

    def select_population(
        self,
        population,
        clones
    ):

        combined = np.vstack(
            [
                population,
                clones
            ]
        )

        fitness_values = self.evaluate_population(
            combined
        )

        sorted_indices = np.argsort(
            fitness_values
        )

        selected = combined[
            sorted_indices[
                :self.config.population_size
            ]
        ]

        return selected


    # ----------------------------------------------------------
    # RANDOM ANTIBODY INJECTION
    # ----------------------------------------------------------

    def inject_random_antibodies(
        self,
        population
    ):

        count = min(
            self.config.random_injection,
            len(population) - 1
        )

        if count <= 0:

            return population

        random_antibodies = self.rng.normal(
            loc=0.0,
            scale=1.0,
            size=(
                count,
                NUM_PARAMETERS
            )
        )

        # Replace worst antibodies
        # while preserving the best.

        population[-count:] = (
            random_antibodies
        )

        return population


    # ----------------------------------------------------------
    # TRAIN AIS
    # ----------------------------------------------------------

    def fit(self):

        print("\n" + "=" * 65)
        print("TRAINING ARTIFICIAL IMMUNE SYSTEM")
        print("=" * 65)

        population = (
            self.initialize_population()
        )

        for generation in range(
            self.config.generations
        ):

            # Evaluate current antibodies
            fitness_values = (
                self.evaluate_population(
                    population
                )
            )

            # Sort by fitness
            sorted_indices = np.argsort(
                fitness_values
            )

            population = population[
                sorted_indices
            ]

            fitness_values = fitness_values[
                sorted_indices
            ]

            # Select elite antibodies
            elites = population[
                :self.config.elite_size
            ]

            elite_fitness = fitness_values[
                :self.config.elite_size
            ]

            # Clone elite antibodies
            clones = self.clone_antibodies(
                elites
            )

            # Apply hypermutation
            mutated_clones = self.hypermutate(
                clones,
                elite_fitness,
                generation
            )

            # Select best antibodies
            population = self.select_population(
                population,
                mutated_clones
            )

            # Preserve best antibody
            current_fitness = fitness_function(
                population[0]
            )

            if current_fitness < self.best_fitness:

                self.best_fitness = (
                    current_fitness
                )

                self.best_antibody = (
                    population[0].copy()
                )

            # Record optimization history
            self.history.append(
                {
                    "Generation":
                        generation + 1,

                    "Best_Fitness":
                        self.best_fitness,

                    "Current_Best_Fitness":
                        current_fitness
                }
            )

            # Random antibody injection
            if generation < (
                self.config.generations - 1
            ):

                population = (
                    self.inject_random_antibodies(
                        population
                    )
                )

            # Display progress
            if (
                generation + 1
            ) % 50 == 0:

                print(
                    f"Generation {generation + 1}"
                    f" | Best Fitness: "
                    f"{self.best_fitness:.8f}"
                )

        print("\nAIS Training Completed.")

        print(
            "Best Fitness:",
            self.best_fitness
        )

        return self


# ==============================================================
# 14. TRAIN AIS MODEL
# ==============================================================

ais = ArtificialImmuneSystem(
    config
)

ais.fit()


# ==============================================================
# 15. CREATE AIS REGRESSION MODEL
# ==============================================================

class AISRegressionModel:

    def __init__(
        self,
        coefficients,
        base_year,
        degree,
        target_mean,
        target_std
    ):

        self.coefficients = np.asarray(
            coefficients,
            dtype=float
        )

        self.base_year = int(
            base_year
        )

        self.degree = int(
            degree
        )

        self.target_mean = np.asarray(
            target_mean,
            dtype=float
        )

        self.target_std = np.asarray(
            target_std,
            dtype=float
        )


    def predict(self, years):

        years = np.asarray(
            years,
            dtype=float
        ).reshape(-1)

        normalized_years = (
            years - self.base_year
        )

        features = polynomial_features(
            normalized_years,
            self.degree
        )

        scaled_predictions = (
            features @ self.coefficients
        )

        predictions = (
            scaled_predictions
            * self.target_std
            + self.target_mean
        )

        return predictions


coefficients = decode_antibody(
    ais.best_antibody
)

ais_model = AISRegressionModel(
    coefficients=coefficients,
    base_year=BASE_YEAR,
    degree=config.polynomial_degree,
    target_mean=y_mean,
    target_std=y_std
)


# ==============================================================
# 16. SAVE AIS MODEL AS PKL
# ==============================================================

print("\nSaving AIS Model as PKL...")

joblib.dump(
    ais_model,
    output_path("model.pkl")
)

print(
    "[SAVED]",
    output_path("model.pkl")
)


# ==============================================================
# 17. SAVE AIS MODEL AS H5
# ==============================================================

# The H5 file stores the numerical parameters
# of the AIS regression model.
#
# It is not a Keras neural network file.

print("\nSaving AIS Model as H5...")

with h5py.File(
    output_path("model.h5"),
    "w"
) as h5:

    h5.create_dataset(
        "coefficients",
        data=coefficients
    )

    h5.create_dataset(
        "target_mean",
        data=y_mean
    )

    h5.create_dataset(
        "target_std",
        data=y_std
    )

    h5.attrs["base_year"] = (
        BASE_YEAR
    )

    h5.attrs["polynomial_degree"] = (
        config.polynomial_degree
    )

    h5.attrs["algorithm"] = (
        "Artificial Immune System - CLONALG"
    )

print(
    "[SAVED]",
    output_path("model.h5")
)


# ==============================================================
# 18. GENERATE HISTORICAL PREDICTIONS
# ==============================================================

ais_predictions = ais_model.predict(
    X.flatten()
)

print("\nAIS Historical Predictions:")
print(ais_predictions)


# ==============================================================
# 19. CALCULATE AIS TRAINING METRICS
# ==============================================================

print("\nCalculating AIS Training Metrics...")

metrics_records = []

for i, target in enumerate(target_names):

    actual = y[:, i]

    predicted = ais_predictions[:, i]

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(
        mse
    )

    r2 = r2_score(
        actual,
        predicted
    )

    metrics_records.append(
        {
            "Algorithm": "AIS-CLONALG",

            "Target": target,

            "MAE": float(mae),

            "MSE": float(mse),

            "RMSE": float(rmse),

            "R2_Score": float(r2)
        }
    )

metrics_df = pd.DataFrame(
    metrics_records
)

print("\nAIS Training Metrics:")
print(metrics_df)

metrics_df.to_csv(
    output_path("metrics.csv"),
    index=False
)


# ==============================================================
# 20. AIS ACCURACY GRAPH
# ==============================================================

print("\nGenerating AIS Accuracy Graph...")

# R² is a regression goodness-of-fit measure.
# It is not classification accuracy.
#
# With three records, these are training scores.

plt.figure(
    figsize=(11, 6)
)

bars = plt.bar(
    metrics_df["Target"],
    metrics_df["R2_Score"],
    color=["steelblue", "darkorange"]
)

plt.title(
    "AIS Mining Model Training Fit (R²)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category"
)

plt.ylabel(
    "Training R² Score"
)

plt.axhline(
    0,
    color="black",
    linestyle="--"
)

for bar, value in zip(
    bars,
    metrics_df["R2_Score"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.grid(
    axis="y",
    alpha=0.3
)

save_show_graph(
    "accuracy_graph.png"
)


# ==============================================================
# 21. AIS CORRELATION HEATMAP
# ==============================================================

print("\nGenerating AIS Heatmap...")

heatmap_data = df[
    [
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines",
        "Five_Star_Percentage"
    ]
]

correlation_matrix = (
    heatmap_data.corr()
)

plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "AIS Mining Performance Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.yticks(
    rotation=0
)

save_show_graph(
    "heatmap.png"
)


# ==============================================================
# 22. AIS COMPARISON GRAPH
# ==============================================================

print("\nGenerating AIS Comparison Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

for i, target in enumerate(target_names):

    axes[i].plot(
        years,
        y[:, i],
        marker="o",
        linewidth=3,
        label="Actual"
    )

    axes[i].plot(
        years,
        ais_predictions[:, i],
        marker="s",
        linestyle="--",
        linewidth=2,
        label="AIS Predicted"
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "Actual vs AIS Predicted Mining Performance",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "comparison_graph.png"
)


# ==============================================================
# 23. GENERATE AIS RESULT CSV
# ==============================================================

print("\nGenerating AIS Result CSV...")

result_df = pd.DataFrame(
    {
        "Year": df["Year"],

        "Actual_Three_Star_Mines":
            y[:, 0].astype(int),

        "AIS_Three_Star_Prediction":
            ais_predictions[:, 0],

        "Actual_Five_Star_Mines":
            y[:, 1].astype(int),

        "AIS_Five_Star_Prediction":
            ais_predictions[:, 1]
    }
)

result_df["AIS_Three_Star_Error"] = np.abs(
    result_df["Actual_Three_Star_Mines"]
    - result_df["AIS_Three_Star_Prediction"]
)

result_df["AIS_Five_Star_Error"] = np.abs(
    result_df["Actual_Five_Star_Mines"]
    - result_df["AIS_Five_Star_Prediction"]
)

result_df.to_csv(
    output_path("result.csv"),
    index=False
)

print("\nAIS Result CSV:")
print(result_df)


# ==============================================================
# 24. AIS RESULT GRAPH
# ==============================================================

print("\nGenerating AIS Result Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

for i, target in enumerate(target_names):

    axes[i].bar(
        np.arange(len(years)) - 0.2,
        y[:, i],
        width=0.4,
        label="Actual"
    )

    axes[i].bar(
        np.arange(len(years)) + 0.2,
        ais_predictions[:, i],
        width=0.4,
        label="AIS Predicted"
    )

    axes[i].set_xticks(
        np.arange(len(years))
    )

    axes[i].set_xticklabels(
        years
    )

    axes[i].set_title(
        target
    )

    axes[i].set_xlabel(
        "Financial Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

fig.suptitle(
    "AIS Mining Performance Results",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "result_graph.png"
)


# ==============================================================
# 25. GENERATE FUTURE PREDICTIONS
# ==============================================================

print("\nGenerating AIS Future Predictions...")

last_year = int(
    df["Year_Numeric"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
)

future_predictions_raw = ais_model.predict(
    future_years
)


# ==============================================================
# 26. CLEAN FUTURE PREDICTIONS
# ==============================================================

def clean_predictions(predictions):

    predictions = np.asarray(
        predictions,
        dtype=float
    ).copy()

    # Prevent negative mine counts
    predictions = np.maximum(
        predictions,
        0
    )

    # Round to integer mine counts
    predictions = np.rint(
        predictions
    ).astype(int)

    # Five-star mines are a subset
    # of three-star-and-above mines.

    predictions[:, 1] = np.minimum(
        predictions[:, 1],
        predictions[:, 0]
    )

    return predictions


future_predictions = clean_predictions(
    future_predictions_raw
)


# ==============================================================
# 27. CREATE AIS PREDICTION CSV
# ==============================================================

future_labels = [
    f"{year}-{str((year + 1) % 100).zfill(2)}"
    for year in future_years
]

prediction_df = pd.DataFrame(
    {
        "Year": future_labels,

        "AIS_Three_Star_Mines":
            future_predictions[:, 0],

        "AIS_Five_Star_Mines":
            future_predictions[:, 1]
    }
)

prediction_df["AIS_Five_Star_Percentage"] = (
    prediction_df["AIS_Five_Star_Mines"]
    / prediction_df[
        "AIS_Three_Star_Mines"
    ].replace(0, np.nan)
) * 100

prediction_df[
    "AIS_Five_Star_Percentage"
] = prediction_df[
    "AIS_Five_Star_Percentage"
].fillna(0)

prediction_df.to_csv(
    output_path("prediction.csv"),
    index=False
)

print("\nAIS Future Predictions:")
print(prediction_df)


# ==============================================================
# 28. AIS PREDICTION GRAPH
# ==============================================================

print("\nGenerating AIS Prediction Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(17, 7)
)

historical_years = (
    df["Year_Numeric"].values
)

for i, target in enumerate(target_names):

    # Historical actual values
    axes[i].plot(
        historical_years,
        y[:, i],
        marker="o",
        linewidth=3,
        label="Historical Actual"
    )

    # Future AIS predictions
    axes[i].plot(
        future_years,
        future_predictions[:, i],
        marker="s",
        linestyle="--",
        linewidth=2,
        label="AIS Forecast"
    )

    # Connect historical data and forecasts
    axes[i].plot(
        [
            historical_years[-1],
            future_years[0]
        ],
        [
            y[-1, i],
            future_predictions[0, i]
        ],
        linestyle="--",
        alpha=0.5
    )

    # Forecast boundary
    axes[i].axvline(
        historical_years[-1],
        linestyle=":",
        color="gray",
        label="Forecast Boundary"
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year Starting Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "AIS Mining Performance Future Forecast "
    "(Illustrative Only)",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "prediction_graph.png"
)


# ==============================================================
# 29. AIS ERROR GRAPH
# ==============================================================

print("\nGenerating AIS Error Graph...")

plt.figure(
    figsize=(11, 6)
)

sns.barplot(
    data=metrics_df,
    x="Target",
    y="MAE",
    color="teal"
)

plt.title(
    "AIS Training Mean Absolute Error",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category"
)

plt.ylabel(
    "Mean Absolute Error (Mine Count)"
)

plt.grid(
    axis="y",
    alpha=0.3
)

save_show_graph(
    "error_graph.png"
)


# ==============================================================
# 30. AIS CONVERGENCE GRAPH
# ==============================================================

print("\nGenerating AIS Convergence Graph...")

history_df = pd.DataFrame(
    ais.history
)

history_df.to_csv(
    output_path("training_history.csv"),
    index=False
)

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    history_df["Generation"],
    history_df["Best_Fitness"],
    linewidth=2,
    label="Best AIS Fitness"
)

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Generation"
)

plt.ylabel(
    "Fitness (Lower Is Better)"
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_show_graph(
    "convergence_graph.png"
)


# ==============================================================
# 31. SAVE AIS YAML CONFIGURATION
# ==============================================================

print("\nSaving AIS Configuration...")

yaml_config = {

    "project": {
        "name":
            "AI-Based Mining Performance Rating Analysis",

        "algorithm":
            "Artificial Immune System",

        "method":
            "Clonal Selection Algorithm",

        "dataset":
            os.path.basename(DATASET_PATH)
    },

    "ais_parameters":
        asdict(config),

    "model": {
        "type":
            "Multi-Output Linear Regression",

        "optimization":
            "AIS-CLONALG",

        "model_file":
            "ais_model.pkl",

        "h5_file":
            "ais_model.h5"
    },

    "dataset": {
        "records":
            int(len(df)),

        "features": [
            "Year_Numeric"
        ],

        "targets": [
            "Three_Star_Mines",
            "Five_Star_Mines"
        ]
    },

    "forecast": {
        "years": 5,

        "file":
            "ais_prediction.csv"
    },

    "evaluation": {
        "type":
            "In-sample training evaluation",

        "validated_test_accuracy":
            False
    }
}

with open(
    output_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False
    )


# ==============================================================
# 32. SAVE AIS JSON METADATA
# ==============================================================

print("\nSaving AIS Metadata...")

metadata = {

    "project_name":
        "AI-Based Mining Performance Rating Analysis",

    "algorithm":
        "Artificial Immune System",

    "optimization_method":
        "Clonal Selection Algorithm",

    "dataset_records":
        int(len(df)),

    "year_range": {
        "start":
            int(df["Year_Numeric"].min()),

        "end":
            int(df["Year_Numeric"].max())
    },

    "ais_parameters":
        asdict(config),

    "best_fitness":
        float(ais.best_fitness),

    "best_antibody":
        ais.best_antibody.tolist(),

    "model_coefficients":
        coefficients.tolist(),

    "training_metrics":
        metrics_df.to_dict(
            orient="records"
        ),

    "future_predictions":
        prediction_df.to_dict(
            orient="records"
        ),

    "generated_files": [
        "ais_model.pkl",
        "ais_model.h5",
        "ais_accuracy_graph.png",
        "ais_heatmap.png",
        "ais_comparison_graph.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_error_graph.png",
        "ais_convergence_graph.png"
    ],

    "limitations":
        (
            "Only three annual observations are available. "
            "The reported metrics are training-fit metrics. "
            "Future forecasts are illustrative extrapolations."
        )
}

with open(
    output_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )


# ==============================================================
# 33. VERIFY SAVED AIS MODEL
# ==============================================================

print("\nVerifying Saved AIS Model...")

loaded_model = joblib.load(
    output_path("model.pkl")
)

test_prediction = loaded_model.predict(
    [2027]
)

print(
    "\nTest Prediction for Financial Year 2027-28:"
)

print(
    clean_predictions(test_prediction)
)


# ==============================================================
# 34. VERIFY ALL GENERATED FILES
# ==============================================================

print("\n" + "=" * 65)
print("VERIFYING AIS OUTPUT FILES")
print("=" * 65)

generated_files = [

    "model.pkl",
    "model.h5",

    "config.yaml",
    "metadata.json",

    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",

    "result.csv",
    "result_graph.png",

    "prediction.csv",
    "prediction_graph.png",

    "error_graph.png",
    "convergence_graph.png",

    "training_history.csv",
    "metrics.csv",

    "processed_data.csv"
]

for filename in generated_files:

    full_filename = (
        "ais_" + filename
    )

    path = os.path.join(
        BASE_DIR,
        full_filename
    )

    if os.path.exists(path):

        print(
            "[SUCCESS]",
            full_filename
        )

    else:

        print(
            "[FAILED]",
            full_filename
        )


# ==============================================================
# 35. FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 70)

print(
    "AI-BASED MINING PERFORMANCE RATING ANALYSIS"
)

print(
    "USING ARTIFICIAL IMMUNE SYSTEM"
)

print("=" * 70)

print("\nAlgorithm:")
print("Artificial Immune System")

print("\nOptimization:")
print("Clonal Selection Algorithm (CLONALG)")

print("\nAIS Parameters:")

print(
    "Population Size:",
    config.population_size
)

print(
    "Generations:",
    config.generations
)

print(
    "Elite Size:",
    config.elite_size
)

print(
    "Clones Per Antibody:",
    config.clones_per_antibody
)

print(
    "Mutation Rate:",
    config.mutation_rate
)

print("\nBest Fitness:")
print(ais.best_fitness)

print("\nGenerated Graphs:")

print("1. ais_accuracy_graph.png")
print("2. ais_heatmap.png")
print("3. ais_comparison_graph.png")
print("4. ais_result_graph.png")
print("5. ais_prediction_graph.png")
print("6. ais_error_graph.png")
print("7. ais_convergence_graph.png")

print("\nGenerated CSV Files:")

print("1. ais_result.csv")
print("2. ais_prediction.csv")
print("3. ais_metrics.csv")
print("4. ais_training_history.csv")
print("5. ais_processed_data.csv")

print("\nGenerated Model Files:")

print("1. ais_model.pkl")
print("2. ais_model.h5")

print("\nGenerated Configuration Files:")

print("1. ais_config.yaml")
print("2. ais_metadata.json")

print("\nOutput Directory:")
print(BASE_DIR)

print(
    "\nWARNING: The dataset contains very few "
    "observations. Results are for educational "
    "demonstration, not validated forecasting."
)

print("\nAIS PROJECT EXECUTION COMPLETED SUCCESSFULLY!")

print("=" * 70)
```
