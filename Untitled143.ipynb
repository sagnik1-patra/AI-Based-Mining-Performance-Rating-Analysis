```python
import os
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam


warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)

sns.set_theme(style="whitegrid")


# ==============================================================
# 2. DEFINE PROJECT PATH
# ==============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Mining Performance Rating Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_265_AU_1557_A_to_D.csv"
)

os.makedirs(
    BASE_DIR,
    exist_ok=True
)


# ==============================================================
# 3. HELPER FUNCTION TO SAVE AND DISPLAY GRAPHS
# ==============================================================

def save_show_graph(filename):

    path = os.path.join(
        BASE_DIR,
        filename
    )

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        f"\nGraph saved successfully: {path}"
    )

    plt.show()

    plt.close()


# ==============================================================
# 4. LOAD DATASET
# ==============================================================

print("\n" + "=" * 60)
print("LOADING MINING PERFORMANCE DATASET")
print("=" * 60)

df = pd.read_csv(
    DATASET_PATH
)

# Remove unnecessary whitespace from column names
df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())


# ==============================================================
# 5. DATA PREPROCESSING
# ==============================================================

print("\nPerforming Data Preprocessing...")

# Use the first three columns from the dataset
# Column 1 = Financial Year
# Column 2 = Three-star and above mines
# Column 3 = Five-star mines

if df.shape[1] < 3:

    raise ValueError(
        "Dataset must contain at least three columns."
    )

df = df.iloc[:, :3].copy()

df.columns = [
    "Year",
    "Three_Star_Mines",
    "Five_Star_Mines"
]

# Convert numerical columns
df["Three_Star_Mines"] = pd.to_numeric(
    df["Three_Star_Mines"],
    errors="coerce"
)

df["Five_Star_Mines"] = pd.to_numeric(
    df["Five_Star_Mines"],
    errors="coerce"
)

# Extract starting year from financial year
df["Year_Numeric"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year_Numeric"] = pd.to_numeric(
    df["Year_Numeric"],
    errors="coerce"
)

# Remove incomplete records
df = df.dropna(
    subset=[
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
).copy()

df["Year_Numeric"] = (
    df["Year_Numeric"].astype(int)
)

# Sort by year
df = df.sort_values(
    "Year_Numeric"
).reset_index(drop=True)

# Validate dataset
if len(df) < 3:

    raise ValueError(
        "At least three valid annual records "
        "are required for this demonstration."
    )

if df["Year_Numeric"].duplicated().any():

    raise ValueError(
        "Duplicate financial years found."
    )

if (
    (df["Three_Star_Mines"] < 0).any()
    or (df["Five_Star_Mines"] < 0).any()
):

    raise ValueError(
        "Mine counts cannot be negative."
    )

if (
    df["Five_Star_Mines"]
    > df["Three_Star_Mines"]
).any():

    raise ValueError(
        "Five-star mine counts cannot exceed "
        "three-star-and-above mine counts."
    )

# Calculate five-star percentage
df["Five_Star_Percentage"] = (
    df["Five_Star_Mines"]
    / df["Three_Star_Mines"].replace(0, np.nan)
) * 100

df["Five_Star_Percentage"] = (
    df["Five_Star_Percentage"].fillna(0)
)

print("\nProcessed Dataset:")
print(df)


# ==============================================================
# 6. PREPARE INPUT FEATURES AND TARGETS
# ==============================================================

X = df[
    ["Year_Numeric"]
].values.astype(float)

y = df[
    [
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
].values.astype(float)

years = df["Year"].astype(str).values

target_names = [
    "Three-Star and Above Mines",
    "Five-Star Mines"
]


# ==============================================================
# 7. FEATURE SCALING
# ==============================================================

print("\nScaling Dataset...")

x_scaler = StandardScaler()

y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)

y_scaled = y_scaler.fit_transform(y)

joblib.dump(
    x_scaler,
    os.path.join(
        BASE_DIR,
        "x_scaler.pkl"
    )
)

joblib.dump(
    y_scaler,
    os.path.join(
        BASE_DIR,
        "y_scaler.pkl"
    )
)


# ==============================================================
# 8. TRAIN LINEAR REGRESSION MODEL
# ==============================================================

print("\nTraining Linear Regression Model...")

lr_model = LinearRegression()

lr_model.fit(
    X,
    y
)

lr_predictions = lr_model.predict(
    X
)

joblib.dump(
    lr_model,
    os.path.join(
        BASE_DIR,
        "mining_model.pkl"
    )
)

print("Linear Regression Training Completed.")


# ==============================================================
# 9. BUILD ARTIFICIAL NEURAL NETWORK
# ==============================================================

print("\nBuilding Neural Network...")

nn_model = Sequential(
    [
        Input(shape=(1,)),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            8,
            activation="relu"
        ),

        Dense(
            2,
            activation="linear"
        )
    ]
)

nn_model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=["mae"]
)

nn_model.summary()


# ==============================================================
# 10. TRAIN NEURAL NETWORK
# ==============================================================

print("\nTraining Neural Network...")

history = nn_model.fit(
    X_scaled,
    y_scaled,

    epochs=300,

    batch_size=1,

    verbose=0,

    shuffle=False
)

nn_model.save(
    os.path.join(
        BASE_DIR,
        "mining_model.h5"
    )
)

print("Neural Network Training Completed.")


# ==============================================================
# 11. GENERATE NEURAL NETWORK PREDICTIONS
# ==============================================================

nn_predictions_scaled = nn_model.predict(
    X_scaled,
    verbose=0
)

nn_predictions = y_scaler.inverse_transform(
    nn_predictions_scaled
)


# ==============================================================
# 12. CALCULATE MODEL PERFORMANCE
# ==============================================================

print("\nCalculating Model Training Performance...")

models = {
    "Linear Regression": lr_predictions,
    "Neural Network": nn_predictions
}

metrics_records = []

for model_name, predictions in models.items():

    for i, target in enumerate(target_names):

        actual = y[:, i]

        predicted = predictions[:, i]

        mae = mean_absolute_error(
            actual,
            predicted
        )

        mse = mean_squared_error(
            actual,
            predicted
        )

        rmse = np.sqrt(mse)

        r2 = r2_score(
            actual,
            predicted
        )

        metrics_records.append(
            {
                "Model": model_name,

                "Target": target,

                "MAE": mae,

                "MSE": mse,

                "RMSE": rmse,

                "R2_Score": r2
            }
        )

metrics_df = pd.DataFrame(
    metrics_records
)

print("\nTraining Performance Metrics:")
print(metrics_df)

metrics_path = os.path.join(
    BASE_DIR,
    "comparison_metrics.csv"
)

metrics_df.to_csv(
    metrics_path,
    index=False
)

print("\nComparison Metrics CSV Saved.")

print(
    "\nNOTE: These are in-sample training metrics. "
    "They are not independent test accuracy."
)


# ==============================================================
# 13. ACCURACY GRAPH
# ==============================================================

print("\nGenerating Accuracy Graph...")

# This graph displays training-fit R2 values.
# R2 is not classification accuracy.

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=metrics_df,

    x="Target",

    y="R2_Score",

    hue="Model",

    palette="viridis"
)

plt.title(
    "Model Training Fit Comparison (R²)",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category",
    fontsize=12
)

plt.ylabel(
    "Training R² Score",
    fontsize=12
)

plt.axhline(
    y=0,
    color="black",
    linestyle="--",
    linewidth=1
)

plt.legend(
    title="Machine Learning Models"
)

plt.grid(
    axis="y",
    alpha=0.3
)

save_show_graph(
    "accuracy_graph.png"
)


# ==============================================================
# 14. CORRELATION HEATMAP
# ==============================================================

print("\nGenerating Correlation Heatmap...")

heatmap_data = df[
    [
        "Year_Numeric",
        "Three_Star_Mines",
        "Five_Star_Mines",
        "Five_Star_Percentage"
    ]
]

correlation_matrix = heatmap_data.corr()

plt.figure(
    figsize=(10, 7)
)

sns.heatmap(
    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    linewidths=0.5,

    square=True,

    cbar=True
)

plt.title(
    "Mining Performance Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.yticks(
    rotation=0
)

save_show_graph(
    "heatmap.png"
)


# ==============================================================
# 15. MODEL COMPARISON GRAPH
# ==============================================================

print("\nGenerating Model Comparison Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

for i, target in enumerate(target_names):

    axes[i].plot(
        years,
        y[:, i],

        marker="o",
        linewidth=2.5,

        label="Actual"
    )

    axes[i].plot(
        years,
        lr_predictions[:, i],

        marker="s",
        linestyle="--",

        linewidth=2,

        label="Linear Regression"
    )

    axes[i].plot(
        years,
        nn_predictions[:, i],

        marker="^",
        linestyle=":",

        linewidth=2,

        label="Neural Network"
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "Linear Regression vs Neural Network",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "comparison_graph.png"
)


# ==============================================================
# 16. GENERATE RESULT CSV
# ==============================================================

print("\nGenerating Result CSV...")

result_df = pd.DataFrame(
    {
        "Year": df["Year"],

        "Actual_Three_Star_Mines":
            y[:, 0].astype(int),

        "LR_Three_Star_Prediction":
            lr_predictions[:, 0],

        "NN_Three_Star_Prediction":
            nn_predictions[:, 0],

        "Actual_Five_Star_Mines":
            y[:, 1].astype(int),

        "LR_Five_Star_Prediction":
            lr_predictions[:, 1],

        "NN_Five_Star_Prediction":
            nn_predictions[:, 1]
    }
)

# Calculate absolute prediction errors
result_df["LR_Three_Star_Error"] = np.abs(
    result_df["Actual_Three_Star_Mines"]
    - result_df["LR_Three_Star_Prediction"]
)

result_df["NN_Three_Star_Error"] = np.abs(
    result_df["Actual_Three_Star_Mines"]
    - result_df["NN_Three_Star_Prediction"]
)

result_df["LR_Five_Star_Error"] = np.abs(
    result_df["Actual_Five_Star_Mines"]
    - result_df["LR_Five_Star_Prediction"]
)

result_df["NN_Five_Star_Error"] = np.abs(
    result_df["Actual_Five_Star_Mines"]
    - result_df["NN_Five_Star_Prediction"]
)

result_path = os.path.join(
    BASE_DIR,
    "result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("\nResult CSV:")
print(result_df)

print("\nResult CSV Saved Successfully.")


# ==============================================================
# 17. RESULT GRAPH
# ==============================================================

print("\nGenerating Result Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(16, 6)
)

# Three-star mines
axes[0].plot(
    years,
    y[:, 0],

    marker="o",
    linewidth=3,

    label="Actual"
)

axes[0].plot(
    years,
    lr_predictions[:, 0],

    marker="s",
    linestyle="--",

    label="Linear Regression"
)

axes[0].plot(
    years,
    nn_predictions[:, 0],

    marker="^",
    linestyle=":",

    label="Neural Network"
)

axes[0].set_title(
    "Three-Star and Above Mining Results"
)

axes[0].set_xlabel(
    "Financial Year"
)

axes[0].set_ylabel(
    "Number of Mines"
)

axes[0].legend()


# Five-star mines
axes[1].plot(
    years,
    y[:, 1],

    marker="o",
    linewidth=3,

    label="Actual"
)

axes[1].plot(
    years,
    lr_predictions[:, 1],

    marker="s",
    linestyle="--",

    label="Linear Regression"
)

axes[1].plot(
    years,
    nn_predictions[:, 1],

    marker="^",
    linestyle=":",

    label="Neural Network"
)

axes[1].set_title(
    "Five-Star Mining Results"
)

axes[1].set_xlabel(
    "Financial Year"
)

axes[1].set_ylabel(
    "Number of Mines"
)

axes[1].legend()

fig.suptitle(
    "Actual vs Predicted Mining Performance",
    fontsize=16,
    fontweight="bold"
)

save_show_graph(
    "result_graph.png"
)


# ==============================================================
# 18. GENERATE FUTURE PREDICTIONS
# ==============================================================

print("\nGenerating Future Mining Predictions...")

last_year = int(
    df["Year_Numeric"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
)

future_X = future_years.reshape(
    -1,
    1
).astype(float)


# Linear Regression future predictions
future_lr = lr_model.predict(
    future_X
)


# Neural Network future predictions
future_X_scaled = x_scaler.transform(
    future_X
)

future_nn_scaled = nn_model.predict(
    future_X_scaled,
    verbose=0
)

future_nn = y_scaler.inverse_transform(
    future_nn_scaled
)


# ==============================================================
# 19. CLEAN FUTURE PREDICTIONS
# ==============================================================

def clean_predictions(predictions):

    predictions = np.asarray(
        predictions,
        dtype=float
    ).copy()

    # Prevent negative predictions
    predictions = np.maximum(
        predictions,
        0
    )

    # Round to whole mine counts
    predictions = np.rint(
        predictions
    ).astype(int)

    # Five-star mines must be a subset
    # of three-star-and-above mines
    predictions[:, 1] = np.minimum(
        predictions[:, 1],
        predictions[:, 0]
    )

    return predictions


future_lr = clean_predictions(
    future_lr
)

future_nn = clean_predictions(
    future_nn
)


# ==============================================================
# 20. CREATE PREDICTION CSV
# ==============================================================

future_labels = [
    f"{year}-{str((year + 1) % 100).zfill(2)}"
    for year in future_years
]

prediction_df = pd.DataFrame(
    {
        "Year": future_labels,

        "LR_Three_Star_Mines":
            future_lr[:, 0],

        "NN_Three_Star_Mines":
            future_nn[:, 0],

        "LR_Five_Star_Mines":
            future_lr[:, 1],

        "NN_Five_Star_Mines":
            future_nn[:, 1]
    }
)

prediction_path = os.path.join(
    BASE_DIR,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("\nFuture Prediction Results:")
print(prediction_df)

print("\nPrediction CSV Saved Successfully.")


# ==============================================================
# 21. PREDICTION GRAPH
# ==============================================================

print("\nGenerating Prediction Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(17, 7)
)

historical_years = (
    df["Year_Numeric"].values
)

for i, target in enumerate(target_names):

    # Historical values
    axes[i].plot(
        historical_years,
        y[:, i],

        marker="o",
        linewidth=3,

        label="Historical Actual"
    )

    # Linear Regression forecast
    axes[i].plot(
        future_years,
        future_lr[:, i],

        marker="s",
        linestyle="--",

        linewidth=2,

        label="Linear Regression Forecast"
    )

    # Neural Network forecast
    axes[i].plot(
        future_years,
        future_nn[:, i],

        marker="^",
        linestyle=":",

        linewidth=2,

        label="Neural Network Forecast"
    )

    # Connect last historical point
    # with first forecast point
    axes[i].plot(
        [
            historical_years[-1],
            future_years[0]
        ],

        [
            y[-1, i],
            future_lr[0, i]
        ],

        linestyle="--",
        alpha=0.5
    )

    axes[i].plot(
        [
            historical_years[-1],
            future_years[0]
        ],

        [
            y[-1, i],
            future_nn[0, i]
        ],

        linestyle=":",
        alpha=0.5
    )

    # Forecast starting boundary
    axes[i].axvline(
        x=historical_years[-1],

        linestyle="--",
        color="gray",

        alpha=0.7
    )

    axes[i].set_title(
        target,
        fontsize=14,
        fontweight="bold"
    )

    axes[i].set_xlabel(
        "Financial Year Starting Year"
    )

    axes[i].set_ylabel(
        "Number of Mines"
    )

    axes[i].legend()

    axes[i].grid(
        alpha=0.3
    )

fig.suptitle(
    "Mining Performance Future Forecast "
    "(Illustrative Only)",
    fontsize=17,
    fontweight="bold"
)

save_show_graph(
    "prediction_graph.png"
)


# ==============================================================
# 22. ADDITIONAL TRAINING ERROR GRAPH
# ==============================================================

print("\nGenerating Training Error Graph...")

plt.figure(
    figsize=(12, 6)
)

sns.barplot(
    data=metrics_df,

    x="Target",

    y="MAE",

    hue="Model",

    palette="Set2"
)

plt.title(
    "Model Training Mean Absolute Error",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Mining Rating Category"
)

plt.ylabel(
    "Mean Absolute Error (Mine Count)"
)

plt.legend(
    title="Models"
)

save_show_graph(
    "error_graph.png"
)


# ==============================================================
# 23. DISPLAY DATASET STATISTICS
# ==============================================================

print("\n" + "=" * 60)
print("MINING DATASET STATISTICS")
print("=" * 60)

print(
    "\nTotal Records:",
    len(df)
)

print(
    "\nAverage Three-Star Mines:",
    df["Three_Star_Mines"].mean()
)

print(
    "\nAverage Five-Star Mines:",
    df["Five_Star_Mines"].mean()
)

print(
    "\nMaximum Three-Star Mines:",
    df["Three_Star_Mines"].max()
)

print(
    "\nMaximum Five-Star Mines:",
    df["Five_Star_Mines"].max()
)

print(
    "\nAverage Five-Star Percentage:",
    df["Five_Star_Percentage"].mean()
)


# ==============================================================
# 24. VERIFY ALL GENERATED FILES
# ==============================================================

print("\n" + "=" * 60)
print("VERIFYING GENERATED OUTPUT FILES")
print("=" * 60)

generated_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "error_graph.png",
    "comparison_metrics.csv",
    "mining_model.pkl",
    "mining_model.h5",
    "x_scaler.pkl",
    "y_scaler.pkl"
]

for filename in generated_files:

    filepath = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(filepath):

        print(
            "[SUCCESS]",
            filename
        )

    else:

        print(
            "[FAILED]",
            filename
        )


# ==============================================================
# 25. FINAL PROJECT SUMMARY
# ==============================================================

print("\n" + "=" * 65)

print(
    "AI-BASED MINING PERFORMANCE RATING ANALYSIS"
)

print("=" * 65)

print("\nModels Used:")

print("1. Linear Regression")
print("2. Artificial Neural Network")

print("\nGraphs Generated:")

print("1. Accuracy Graph")
print("2. Correlation Heatmap")
print("3. Model Comparison Graph")
print("4. Result Graph")
print("5. Prediction Graph")
print("6. Training Error Graph")

print("\nCSV Files Generated:")

print("1. result.csv")
print("2. prediction.csv")
print("3. comparison_metrics.csv")

print("\nModel Files:")

print("1. mining_model.pkl")
print("2. mining_model.h5")

print("\nOutput Directory:")

print(BASE_DIR)

print(
    "\nIMPORTANT: The dataset has very few "
    "observations. Training-fit scores and "
    "forecasts are demonstration outputs, "
    "not validated real-world performance."
)

print("\nPROJECT EXECUTION COMPLETED SUCCESSFULLY!")

print("=" * 65)
```
