```python
import os
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam

warnings.filterwarnings("ignore")

# ============================================================
# 2. DEFINE PROJECT PATHS
# ============================================================

BASE_DIR = r"C:\Users\sagni\Downloads\AI-Based Mining Performance Rating Analysis"

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_265_AU_1557_A_to_D.csv"
)

H5_PATH = os.path.join(
    BASE_DIR,
    "mining_model.h5"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "mining_model.pkl"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "mining_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "mining_metadata.json"
)

PREDICTION_PATH = os.path.join(
    BASE_DIR,
    "mining_predictions.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

# ============================================================
# 3. SET RANDOM SEEDS
# ============================================================

np.random.seed(42)
tf.random.set_seed(42)

# ============================================================
# 4. LOAD DATASET
# ============================================================

print("\nLoading Mining Dataset...")

df = pd.read_csv(DATASET_PATH)

print("\nOriginal Dataset:")
print(df)

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 5. DATA PREPROCESSING
# ============================================================

print("\nPerforming Data Preprocessing...")

# Original dataset column names
YEAR_COL = "Year"

THREE_STAR_COL = (
    "No. of 3 Star and above Rated Mines (a)"
)

FIVE_STAR_COL = (
    "No. of 5 Star Rated Mines out of (a)"
)

# Rename columns
df = df.rename(
    columns={
        YEAR_COL: "Year",
        THREE_STAR_COL: "Three_Star_Mines",
        FIVE_STAR_COL: "Five_Star_Mines"
    }
)

required_columns = [
    "Year",
    "Three_Star_Mines",
    "Five_Star_Mines"
]

# Verify columns
for col in required_columns:
    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

# Remove missing records
df = df.dropna(subset=required_columns).copy()

# Convert numeric columns
df["Three_Star_Mines"] = pd.to_numeric(
    df["Three_Star_Mines"],
    errors="raise"
)

df["Five_Star_Mines"] = pd.to_numeric(
    df["Five_Star_Mines"],
    errors="raise"
)

# Extract starting year
df["Year_Numeric"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"^(\d{4})")[0]
    .astype(int)
)

# Sort chronologically
df = df.sort_values(
    "Year_Numeric"
).reset_index(drop=True)

# Check for duplicate years
if df["Year_Numeric"].duplicated().any():
    raise ValueError(
        "Duplicate financial years detected."
    )

# Check values
if (
    (df["Three_Star_Mines"] < 0).any()
    or (df["Five_Star_Mines"] < 0).any()
):
    raise ValueError(
        "Mine counts cannot be negative."
    )

if (
    df["Five_Star_Mines"]
    > df["Three_Star_Mines"]
).any():
    raise ValueError(
        "Five-star mines cannot exceed "
        "three-star-and-above mines."
    )

# Calculate five-star percentage
df["Five_Star_Percentage"] = (
    df["Five_Star_Mines"]
    / df["Three_Star_Mines"]
) * 100

# Calculate year-on-year growth
df["Three_Star_Growth"] = (
    df["Three_Star_Mines"]
    .pct_change() * 100
)

df["Five_Star_Growth"] = (
    df["Five_Star_Mines"]
    .pct_change() * 100
)

print("\nProcessed Dataset:")
print(df)

# Save processed dataset
processed_path = os.path.join(
    BASE_DIR,
    "processed_mining_data.csv"
)

df.to_csv(
    processed_path,
    index=False
)

print("\nProcessed dataset saved.")

# ============================================================
# 6. PREPARE FEATURES AND TARGETS
# ============================================================

print("\nPreparing Features and Targets...")

# Input feature: financial year
X = df[["Year_Numeric"]].values.astype(float)

# Output targets
y = df[
    [
        "Three_Star_Mines",
        "Five_Star_Mines"
    ]
].values.astype(float)

# ============================================================
# 7. FEATURE SCALING
# ============================================================

print("\nScaling Features...")

# Separate scalers for input and output
x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)
y_scaled = y_scaler.fit_transform(y)

# Save scalers
joblib.dump(
    x_scaler,
    os.path.join(BASE_DIR, "x_scaler.pkl")
)

joblib.dump(
    y_scaler,
    os.path.join(BASE_DIR, "y_scaler.pkl")
)

print("Scalers saved successfully.")

# ============================================================
# 8. TRAIN LINEAR REGRESSION MODEL
# ============================================================

print("\nTraining Linear Regression Model...")

lr_model = LinearRegression()

lr_model.fit(
    X,
    y
)

# Training predictions
lr_predictions = lr_model.predict(X)

# Save Linear Regression model
joblib.dump(
    lr_model,
    PKL_PATH
)

print("\nLinear Regression Model Saved:")
print(PKL_PATH)

# ============================================================
# 9. BUILD NEURAL NETWORK MODEL
# ============================================================

print("\nBuilding Neural Network Model...")

nn_model = Sequential(
    [
        Input(shape=(1,)),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            8,
            activation="relu"
        ),

        Dense(
            2,
            activation="linear"
        )
    ]
)

nn_model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

print("\nNeural Network Architecture:")
nn_model.summary()

# ============================================================
# 10. TRAIN NEURAL NETWORK
# ============================================================

print("\nTraining Neural Network...")

history = nn_model.fit(
    X_scaled,
    y_scaled,
    epochs=300,
    batch_size=1,
    verbose=0,
    shuffle=False
)

print("\nNeural Network Training Completed.")

# ============================================================
# 11. SAVE NEURAL NETWORK AS H5
# ============================================================

nn_model.save(
    H5_PATH
)

print("\nNeural Network Model Saved:")
print(H5_PATH)

# ============================================================
# 12. GENERATE TRAINING PREDICTIONS
# ============================================================

print("\nGenerating Training Predictions...")

nn_predictions_scaled = nn_model.predict(
    X_scaled,
    verbose=0
)

nn_predictions = y_scaler.inverse_transform(
    nn_predictions_scaled
)

# Calculate descriptive training errors
lr_mae = np.mean(
    np.abs(y - lr_predictions),
    axis=0
)

nn_mae = np.mean(
    np.abs(y - nn_predictions),
    axis=0
)

print("\nLinear Regression Training MAE:")
print("3-Star Mines:", lr_mae[0])
print("5-Star Mines:", lr_mae[1])

print("\nNeural Network Training MAE:")
print("3-Star Mines:", nn_mae[0])
print("5-Star Mines:", nn_mae[1])

print(
    "\nWARNING: These are training errors, "
    "not test accuracy or validated "
    "forecast performance."
)

# ============================================================
# 13. FUTURE YEAR FORECASTING
# ============================================================

print("\nGenerating Future Predictions...")

last_year = int(
    df["Year_Numeric"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
)

future_X = future_years.reshape(-1, 1).astype(float)

# Linear Regression predictions
future_lr = lr_model.predict(
    future_X
)

# Neural Network predictions
future_X_scaled = x_scaler.transform(
    future_X
)

future_nn_scaled = nn_model.predict(
    future_X_scaled,
    verbose=0
)

future_nn = y_scaler.inverse_transform(
    future_nn_scaled
)

# ============================================================
# 14. POSTPROCESS FUTURE PREDICTIONS
# ============================================================

def process_predictions(predictions):
    """
    Convert continuous predictions to
    nonnegative integer mine counts.

    Ensure five-star counts never exceed
    three-star-and-above counts.
    """

    predictions = np.asarray(
        predictions,
        dtype=float
    )

    predictions = np.maximum(
        predictions,
        0
    )

    predictions = np.rint(
        predictions
    ).astype(int)

    predictions[:, 1] = np.minimum(
        predictions[:, 1],
        predictions[:, 0]
    )

    return predictions


future_lr = process_predictions(
    future_lr
)

future_nn = process_predictions(
    future_nn
)

# ============================================================
# 15. CREATE FUTURE PREDICTION DATAFRAME
# ============================================================

future_labels = [
    f"{year}-{str((year + 1) % 100).zfill(2)}"
    for year in future_years
]

prediction_df = pd.DataFrame(
    {
        "Year": future_labels,

        "LR_Three_Star_Mines":
            future_lr[:, 0],

        "LR_Five_Star_Mines":
            future_lr[:, 1],

        "NN_Three_Star_Mines":
            future_nn[:, 0],

        "NN_Five_Star_Mines":
            future_nn[:, 1]
    }
)

prediction_df.to_csv(
    PREDICTION_PATH,
    index=False
)

print("\nIllustrative Future Predictions:")
print(prediction_df)

print("\nPredictions Saved:")
print(PREDICTION_PATH)

# ============================================================
# 16. CREATE YAML CONFIGURATION FILE
# ============================================================

print("\nCreating YAML Configuration...")

config = {

    "project": {
        "name":
            "AI-Based Mining Performance Rating Analysis",

        "domain":
            "Machine Learning",

        "task":
            "Multi-output Regression",

        "dataset":
            os.path.basename(DATASET_PATH)
    },

    "dataset": {
        "rows": int(len(df)),

        "features": [
            "Year_Numeric"
        ],

        "targets": [
            "Three_Star_Mines",
            "Five_Star_Mines"
        ]
    },

    "models": {

        "linear_regression": {
            "type":
                "LinearRegression",

            "file":
                "mining_model.pkl"
        },

        "neural_network": {
            "type":
                "Sequential",

            "hidden_layers": [
                32,
                16,
                8
            ],

            "activation":
                "relu",

            "output_activation":
                "linear",

            "optimizer":
                "Adam",

            "loss":
                "mse",

            "epochs":
                300,

            "batch_size":
                1,

            "file":
                "mining_model.h5"
        }
    },

    "preprocessing": {
        "input_scaler":
            "x_scaler.pkl",

        "target_scaler":
            "y_scaler.pkl"
    },

    "forecast": {
        "future_years": 5,

        "output_file":
            "mining_predictions.csv"
    },

    "limitations": {
        "small_dataset": True,

        "validated_test_accuracy":
            False,

        "forecast_type":
            "Illustrative extrapolation only"
    }
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("\nYAML File Saved:")
print(YAML_PATH)

# ============================================================
# 17. CREATE JSON METADATA FILE
# ============================================================

print("\nCreating JSON Metadata...")

metadata = {

    "project_name":
        "AI-Based Mining Performance Rating Analysis",

    "dataset_name":
        os.path.basename(DATASET_PATH),

    "total_records":
        int(len(df)),

    "year_range": {
        "start":
            int(df["Year_Numeric"].min()),

        "end":
            int(df["Year_Numeric"].max())
    },

    "dataset_statistics": {

        "average_three_star_mines":
            float(df["Three_Star_Mines"].mean()),

        "average_five_star_mines":
            float(df["Five_Star_Mines"].mean()),

        "maximum_three_star_mines":
            int(df["Three_Star_Mines"].max()),

        "maximum_five_star_mines":
            int(df["Five_Star_Mines"].max()),

        "average_five_star_percentage":
            float(
                df["Five_Star_Percentage"].mean()
            )
    },

    "training_errors": {

        "linear_regression": {

            "three_star_mae":
                float(lr_mae[0]),

            "five_star_mae":
                float(lr_mae[1])
        },

        "neural_network": {

            "three_star_mae":
                float(nn_mae[0]),

            "five_star_mae":
                float(nn_mae[1])
        }
    },

    "neural_network": {

        "epochs":
            300,

        "final_training_loss":
            float(history.history["loss"][-1]),

        "final_training_mae_scaled":
            float(history.history["mae"][-1])
    },

    "model_files": {

        "linear_regression":
            "mining_model.pkl",

        "neural_network":
            "mining_model.h5",

        "input_scaler":
            "x_scaler.pkl",

        "target_scaler":
            "y_scaler.pkl"
    },

    "forecast_years":
        future_labels,

    "forecast_results":
        prediction_df.to_dict(
            orient="records"
        ),

    "evaluation_note":
        (
            "Only three observations are available. "
            "Models were trained on all observations. "
            "Reported MAE values are in-sample errors, "
            "not independent test performance. "
            "Forecasts are illustrative and unreliable "
            "for real-world decision-making."
        )
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("\nJSON File Saved:")
print(JSON_PATH)

# ============================================================
# 18. VERIFY SAVED FILES
# ============================================================

print("\nVerifying Saved Files...")

files_to_check = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,
    PREDICTION_PATH
]

for path in files_to_check:

    if os.path.exists(path):

        print(
            "[SUCCESS]",
            os.path.basename(path)
        )

    else:

        print(
            "[FAILED]",
            os.path.basename(path)
        )

# ============================================================
# 19. TEST LOADING SAVED MODELS
# ============================================================

print("\nTesting Saved Models...")

# Load Linear Regression
loaded_lr = joblib.load(
    PKL_PATH
)

# Load Neural Network
loaded_nn = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

# Test input
test_year = np.array(
    [[2027.0]]
)

# Linear Regression prediction
test_lr_prediction = loaded_lr.predict(
    test_year
)

# Neural Network prediction
test_year_scaled = x_scaler.transform(
    test_year
)

test_nn_scaled = loaded_nn.predict(
    test_year_scaled,
    verbose=0
)

test_nn_prediction = y_scaler.inverse_transform(
    test_nn_scaled
)

print("\nTest Year: 2027")

print(
    "Linear Regression Prediction:",
    process_predictions(test_lr_prediction)
)

print(
    "Neural Network Prediction:",
    process_predictions(test_nn_prediction)
)

# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)

print(
    "AI-BASED MINING PERFORMANCE RATING ANALYSIS"
)

print("=" * 60)

print("\nDataset Records:", len(df))

print(
    "Models Trained:",
    "Linear Regression and Neural Network"
)

print("\nGenerated Files:")

for filename in [
    "mining_model.h5",
    "mining_model.pkl",
    "mining_config.yaml",
    "mining_metadata.json",
    "x_scaler.pkl",
    "y_scaler.pkl",
    "processed_mining_data.csv",
    "mining_predictions.csv"
]:

    print(" -", filename)

print("\nOutput Directory:")
print(BASE_DIR)

print(
    "\nNOTE: Forecasts are illustrative "
    "because the dataset contains "
    "only three annual records."
)

print("\nPROJECT EXECUTION COMPLETED SUCCESSFULLY!")

print("=" * 60)
```
